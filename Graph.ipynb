```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial, log2, sqrt

# Input sizes
n = np.arange(1, 21)                    # small range to keep n! visible
n_smooth = np.linspace(1, 20, 400)      # smooth curve for continuous functions

# Continuous functions
O_1      = np.ones_like(n_smooth)
O_logn   = np.log2(n_smooth)
O_sqrtn  = np.sqrt(n_smooth)
O_n      = n_smooth
O_nlogn  = n_smooth * np.log2(n_smooth)
O_n2     = n_smooth ** 2
O_n3     = n_smooth ** 3
O_2n     = 2 ** n_smooth

# Discrete functions (factorial only defined for integers)
O_nfact  = np.array([factorial(int(k)) for k in n])

# ---- Plot ----
plt.figure(figsize=(12, 8))

plt.plot(n_smooth, O_1,     label='O(1)  Constant',      linewidth=2)
plt.plot(n_smooth, O_logn,  label='O(log n)  Logarithmic', linewidth=2)
plt.plot(n_smooth, O_sqrtn, label='O(√n)  Square Root',  linewidth=2)
plt.plot(n_smooth, O_n,     label='O(n)  Linear',        linewidth=2)
plt.plot(n_smooth, O_nlogn, label='O(n log n)  Linearithmic', linewidth=2)
plt.plot(n_smooth, O_n2,    label='O(n²)  Quadratic',    linewidth=2)
plt.plot(n_smooth, O_n3,    label='O(n³)  Cubic',        linewidth=2)
plt.plot(n_smooth, O_2n,    label='O(2ⁿ)  Exponential',  linewidth=2)
plt.plot(n,        O_nfact, label='O(n!)  Factorial',    linewidth=2,
         marker='o', markersize=4)

plt.yscale('log')                       # log scale is essential!
plt.xlabel('Input size  n', fontsize=13)
plt.ylabel('Number of operations (log scale)', fontsize=13)
plt.title('Growth Rates of Common Big-O Complexities', fontsize=15)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.ylim(1, 1e18)                       # cap y-axis so 20! doesn't squash everything
plt.tight_layout()
plt.savefig('bigO_growth.png', dpi=150)
plt.show()
```
